## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？
A:因為 Python 的 int 採用**任意精度整數（Arbitrary Precision Integer）**實作，而不是使用固定大小的機器整數。

2. Python 的 `int` 是否具有固定的位元數限制？
A:沒有

3. 什麼是 Arbitrary Precision（任意精度）？
A:數值可使用任意多個位元儲存，而不是受固定長度限制。

4. CPython 使用什麼物件來實作 `int`？
A:PyLongObject

5. Python 整數大小真正的限制是什麼？
A:沒有固定數值上限

A:能表示的數值大小區間

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？
Python 的 int 不受固定 32 位元或 64 位元限制，可隨數值大小動態增加儲存空間。int 在 CPython 中由 PyLongObject 實作，屬於任意大小整數
2. 它是否代表可以儲存無限大的整數？
不代表
3. CPython 如何實現這項特性？
PyLongObject
---

### 思考題二


比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？
因為 Python 的 int 使用 PyLongObject 實作，屬於「任意大小（arbitrary-size）」整數。當數字變大時，CPython 會配置更多儲存空間來保存額外的數字區塊（digits），而不是限制在固定的 32 位元或 64 位元
2. 為什麼 `float` 卻存在範圍？
Python 的 float 使用 PyFloatObject，底層對應 C 的 double，採用 IEEE 754 雙精度浮點格式
3. 為什麼 `float` 可能產生誤差？
因為許多十進位小數無法用有限長度的二進位表示

---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。

| 項目 | Python `int` | FP16 / BF16 |
|------|-------------|-------------|
| 設計理念 | 正確性優先 | 效率優先 |
| 精度 | 任意精度（Arbitrary Precision） | 有限精度（Limited Precision） |
| 位元數 | 不固定 | 固定 16 位元 |
| 記憶體使用 | 依數值大小增加 | 固定且較小 |
| 運算速度 | 較慢 | 較快 |
| 是否有誤差 | 無（整數可精確表示） | 有（浮點近似值） |
| 是否適合 AI 訓練 | 不適合 | 非常適合 |
| 主要用途 | 整數運算、數學計算 | AI 訓練、AI 推論 |
| 優點 | 結果精確、不易溢位 | 節省記憶體、計算速度快 |
| 缺點 | 記憶體較大、速度較慢 | 存在精度損失與捨入誤差 |

2. 各自的優點與缺點。
Python int
優點: 不會發生整數溢位、可精確表示整數、適合數學與金融等需要正確結果的場景
缺點： 記憶體消耗較大、大整數運算較慢、不適合大量矩陣運算
FP16 / BF16
優點：記憶體需求小、GPU 運算速度快、可訓練更大的模型、降低顯存使用量
缺點： 有捨入誤差、精度較低、某些情況可能造成數值不穩定

3. 為什麼大型語言模型通常選擇較低精度的表示方式？
模型規模太大
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
記憶體大幅增加、訓練速度下降、顯存需求暴增、精度提升有限
