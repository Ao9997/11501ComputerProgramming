0.2 × 2 = 0.4 → 0

得到的二進位為：

0.00011001100110011...
結果
0.1₁₀ = 0.00011001100110011...₂
現象：

0011
會一直重複，因此 0.1 無法用有限位數的二進位精確表示，而是一個無限循環小數

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
由於 0.1 的二進位是無限循環小數：

0.00011001100110011...
Python 採用 IEEE 754 浮點數標準，只能儲存有限位數，因此會將超出的部分以四捨六入五成雙方式取最接近的值。

因此 Python 儲存的是 0.1 的近似值，而非精確值，所以可能出現：

0.1 + 0.2
結果：

0.30000000000000004
這就是浮點數誤差的來源
